```python
import pandas as pd

# 1. Read the input CSV file from Colab workspace
input_filename = 'bangla-med-qa.csv'
df = pd.read_csv(input_filename)

# 2. Define a helper function to replace option letters (A, B, C, D)
# with their actual corresponding text values from the options columns
def get_answer_value(row):
    # Clean the answer letter (e.g., 'C' or 'c' -> 'C')
    ans_letter = str(row['answer']).strip().upper()

    # Construct the matching column name (e.g., 'options/C')
    target_column = f'options/{ans_letter}'

    # If the option column exists and isn't empty, return its actual text value
    if target_column in row and pd.notna(row[target_column]):
        return row[target_column]

    # Fallback in case of missing mapping
    return row['answer']

# 3. Replace the 'answer' column values with the full answer text
df['answer'] = df.apply(get_answer_value, axis=1)

# 4. Keep ONLY 'serial_no', 'question', and the newly updated 'answer' column
# All other columns (options/A-D, exam_name, explanation) are dropped automatically
final_df = df[['serial_no', 'question', 'answer']]

# 5. Save the cleaned DataFrame to a new CSV file
output_filename = 'bangla_med_qa_correct.csv'
final_df.to_csv(output_filename, index=False, encoding='utf-8-sig')

print(f"Processing complete! The cleaned dataset is saved as '{output_filename}'.")
```
