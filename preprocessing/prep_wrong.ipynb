```python
import pandas as pd

# 1. Read the input CSV file from the Colab file tree
input_filename = 'bangla-med-qa.csv'
df = pd.read_csv(input_filename)

# 2. Extract options dynamically and find all incorrect choices for each row
wrong_opt_1 = []
wrong_opt_2 = []
wrong_opt_3 = []

all_letters = ['A', 'B', 'C', 'D']

for _, row in df.iterrows():
    # Get the correct answer key (e.g., 'C')
    correct_key = str(row['answer']).strip().upper()

    # Filter out the correct option letter to get the 3 incorrect letters
    incorrect_keys = [letter for letter in all_letters if letter != correct_key]

    # Grab the corresponding text values for the 3 wrong choices
    wrong_opt_1.append(row[f'options/{incorrect_keys[0]}'])
    wrong_opt_2.append(row[f'options/{incorrect_keys[1]}'])
    wrong_opt_3.append(row[f'options/{incorrect_keys[2]}'])

# 3. Build DataFrames for each of the 3 wrong answer sets
df_wrong_1 = pd.DataFrame({
    'serial_no': df['serial_no'],
    'question': df['question'],
    'answer': wrong_opt_1
})

df_wrong_2 = pd.DataFrame({
    'serial_no': df['serial_no'],
    'question': df['question'],
    'answer': wrong_opt_2
})

df_wrong_3 = pd.DataFrame({
    'serial_no': df['serial_no'],
    'question': df['question'],
    'answer': wrong_opt_3
})

# 4. Save each DataFrame to its own CSV file
df_wrong_1.to_csv('wrong_answers_v1.csv', index=False, encoding='utf-8-sig')
df_wrong_2.to_csv('wrong_answers_v2.csv', index=False, encoding='utf-8-sig')
df_wrong_3.to_csv('wrong_answers_v3.csv', index=False, encoding='utf-8-sig')

print("Done! Created 3 CSV files:")
print("- wrong_answers_v1.csv")
print("- wrong_answers_v2.csv")
print("- wrong_answers_v3.csv")
```
